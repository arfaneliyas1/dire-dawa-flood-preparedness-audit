# Dire Dawa Flood Preparedness Evidence Audit

This notebook documents the reproducible loading, checking, and summary logic for the publication package. It uses only the included CSV compilations and does not load restricted microdata.

```python
from pathlib import Path
import pandas as pd

ROOT = Path('..')
inventory = pd.read_csv(ROOT / 'data' / 'source_inventory.csv')
matrix = pd.read_csv(ROOT / 'data' / 'preparedness_evidence_matrix.csv')
scorecard = pd.read_csv(ROOT / 'data' / 'evidence_quality_scorecard.csv')
inventory.head(), matrix.head(), scorecard.head()
```

## Basic validation

The checks below confirm source references, finding identifiers, and the seven-domain chain.

```python
domains = [
    'Risk monitoring and hazard information',
    'Forecasting and warning generation',
    'Warning dissemination',
    'Warning receipt, understanding, and accessibility',
    'Protective action and evacuation',
    'Shelter, emergency health, and service continuity',
    'Recovery, livelihoods, and learning',
]
assert matrix['finding_id'].is_unique
assert set(matrix['source_id']).issubset(set(inventory['source_id']))
assert set(matrix['domain']) == set(domains)
print('Validation passed:', len(inventory), 'sources and', len(matrix), 'findings')
```
